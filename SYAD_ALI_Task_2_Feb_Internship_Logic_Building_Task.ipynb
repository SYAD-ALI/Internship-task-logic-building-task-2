###TASK-2 Problem 1: Unique Words in a Sentence

```python
#Problem 1: Unique Words in a Sentence
# Take input from user

sentence = input("Enter a sentence: ")
#split the sentence into words

words = sentence.split()
unique_words=set(words)
count=len(unique_words)
print("Number of unique words:",count)
print("Unique words:", unique_words)
```

##Problem 2: Highest Salary from Employee Data

```python
#Problem 2: Highest Salary from Employee Data
#define employee salary

employees={"ALI": 75000,
    "IMAMBI": 600,
    "HAJI": 72000}
#Find employee with highest salary
highest_paid_employee=max(employees,key=employees.get)
highest_salary=employees[highest_paid_employee]

print("Employee with highest salary:",highest_paid_employee)
print(f"Highest Salary: {highest_paid_employee} - {highest_salary}")
```

## TASK-2Problem 3: Find Maximum and Minimum Values

```python
#Problem 3: Find Maximum and Minimum Values
numbers = [10,22,66,77,88]
max = numbers[0]
min = numbers [0]

for i in numbers:
  if i > max:
    max = i
  elif i < min:
    min = i
print("List:", numbers)
print("The max number",max)
print("The minimum number",min)
```

##Problem 4: Count Products Above a Price Threshold

##Problem 5: Calculate Attendance Percentage

```python
#Problem 5: Calculate Attendance Percentage
# Step 1: Define attendance list
attendance = ["present", "absent", "present", "absent", "present", "present", "absent"]
present_count = 0

for status in attendance:
  if status == "present":
    present_count += 1
totle_days = len(attendance)
attendance_percentage = (present_count / totle_days) * 100

print("Attendance Percentage:", attendance_percentage)

rounded_percentage = round(attendance_percentage, 2)
print("Attendance Percentage:", rounded_percentage)
```

##Problem 7: Count Character Frequency

```python
#Problem 7: Count Character Frequency
#Define the string
string = "Hello, World!"
frequency = {}

for char in string:
  if char in frequency:
    frequency[char] += 1
  else:
    frequency[char] = 1
print
print("Character Frequency:", frequency)
```

##Problem 10: Calculate Average Marks


```python
#Problem 10: Calculate Average Marks

#Define list of marks
marks = [70, 75, 80, 65]

#Calculate sum of marks
total = 0
for mark in marks:
    total += mark

#Calculate average
average = total / len(marks)

#Display result
print("Average Marks:", average)
```
